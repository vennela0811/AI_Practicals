```python
room = ["D", "C", "D", "D", "C"]
def show_room(room):
  picture = ""
  for spot in room:
    if spot == "D":
      picture += "💩"
    else:
      picture += "✨"
  print(picture)
print("Our room right now:")
show_room(room)
```

```python
def clean_spot(spot):
  if spot == "D":
    return "C"
  else:
    return "C"
  result = clean_spot("D")
  print("The robot looked at dirty spot 💩 and made it:", result, "(c means clean ✨)")
```

```python
print("BEFORE the dirty room:")
show_room(room)
print()
for i in range(len(room)):
  room[i] = clean_spot(room[i])
print("AFTER the clean spot")
show_room(room)
```

```python
def show_room(room):
  picture = ""
  for spot in room:
    if spot == "D":
      picture += "💩"
    else:
      picture += "✨"
  print(picture)

```

```python
def clean_spot(spot):
  if spot == "D":
    return "C"
  else:
    return "C"
```

```python
room2= ["D", "C", "D", "D", "C"]
cleaned = 0
for i in range(len(room2)):
  if room2[i] == "D":
    cleaned = cleaned + 1
  room2[i] = clean_spot(room2[i])

print("The robot cleaned", cleaned, "dirty spots. 🧹")
print("The room is now: ")
show_room(room2)
```

```python
my_room = ["D", "D", "C", "D", "C", "D", "D", "C", "D"]
print("My room before:")
show_room(my_room)

for i in range(len(my_room)):
  my_room[i] = clean_spot(my_room[i])

print("My room after the robot cleaned it:")
show_room(my_room)
```
